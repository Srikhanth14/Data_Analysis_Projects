# **Data Cleaning and Pre-Processing**

```python
#Converting Date Columns to Date Type
data["Date of Admission"] = pd.to_datetime(data["Date of Admission"])
data["Discharge Date"] = pd.to_datetime(data["Discharge Date"])
```

```python
print(data[["Date of Admission","Discharge Date"]].dtypes)
```

```python
data[["Date of Admission","Discharge Date"]].sample(2)
```

```python
#Admission and Discharge Date Validation
Date_Validation = data['Discharge Date'] > data['Date of Admission']
print("Are all discharge date valid: ", Date_Validation.all())
print("Total Invalid rows in Date_Validation: ",(~Date_Validation).sum())
```

```python
#Length of Stay Validation
Calculated_Length_Of_Stay = (data['Discharge Date'] - data['Date of Admission']).dt.days
Length_Of_Stay_Validation = data['Length of Stay']== Calculated_Length_Of_Stay
print("Does Length of Stay match the Calculated Length_Of_Stay numbers: ", Length_Of_Stay_Validation.all())
print("Total Invalid rows in Length_of_Stay_Validation: ",(~Length_Of_Stay_Validation).sum())
```

```python
#Billing Amount Validation
Billing_Validation = data['Billing Amount']>0
print("Does Billing Amount is greater than zero: ", Billing_Validation.all())
print("Total Invalid rows in Billing Amount: ",(~Billing_Validation).sum())
```

```python
#Age Validation
Age_Validation = data['Age'].between(0,100)
print("Are all ages within the valid ranges: ", Age_Validation.all())
print("Total Invalid Values in Ages: ",(~Age_Validation).sum())
```

```python
#Cross checking the consistency of Categorical Target Columns
Categorical_Target_Columns = ["Gender","Blood Type", "Medical Condition", "Doctor", "Hospital", "Insurance Provider", "Admission Type", "Medication", "Test Results"]

for categorical_columns in Categorical_Target_Columns:
    print(categorical_columns)
    print(data[categorical_columns].unique())
    print("-"*50)
```

# **Feature Engineering**

```python
#Extracting the day, month and year from the Admission Date
data['Day of Admission Week'] = data['Date of Admission'].dt.day_name()
data['Month of Admission'] = data['Date of Admission'].dt.month
data['Year of Admission'] = data['Date of Admission'].dt.year
```

```python
print("Unique values in Admission Year")
print(data["Year of Admission"].unique())
```

```python
print("Unique values in Admission Week Day")
print(data["Day of Admission Week"].unique())
```

```python
print("Unique values in Admission Month")
print(data["Month of Admission"].unique())
```

```python
print("Frequency values in Admission Year", '\n')
print(data.groupby("Year of Admission").size().sort_values(ascending=False))
```

```python
print("Frequency values in Admission Month based on Admission Year", '\n')
print(data.groupby(["Year of Admission","Month of Admission"]).size().sort_values(ascending=False))
```

```python
print("Frequency values in Admission Week Day", '\n')
print(data.groupby(["Year of Admission","Day of Admission Week"]).size().sort_values(ascending=False))
```

#**Feature Encoding**

```python
#Converting the Gender Categorical Variables to Numeric i,e Male := 0, Female := 1
data["Encoded Gender"] = data["Gender"].map({"Male":0,"Female":1})
```
