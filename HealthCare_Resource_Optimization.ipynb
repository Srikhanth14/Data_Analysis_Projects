#**One-Hot Encoding**

```python
#Converting Required Categorical Features to Numeric using One-Hot Encoding
Encoded_Data=pd.get_dummies(data, columns=["Gender","Blood Type","Medical Condition","Hospital","Insurance Provider","Admission Day", "Admission Month","Admission Year"],drop_first=True,dtype=int)
Encoded_Data.info()
```

```python
print(Encoded_Data.columns)
```

# **Label Encoding**

```python
#Encoding the Target Feature by using Label Encoding i,e Mapping Function
Admission_Type_Mapping = {'Emergency': 0, 'Elective': 1, 'Urgent': 2,'Routine':3}
Encoded_Data['Admission Type'] = Encoded_Data['Admission Type'].map(Admission_Type_Mapping)
```

```python
Encoded_Data['Admission Type'].value_counts()
```

```python
Encoded_Data.sample(2)
```

# **Model Selection**

```python
from sklearn.model_selection import train_test_split   #Splitting the Data
from sklearn.linear_model import LogisticRegression    #Logistic Regression
from sklearn.tree import DecisionTreeClassifier        #Decision Tree
from sklearn.ensemble import RandomForestClassifier    #Random Forest Classification
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score  #Model Evaluation Parameters
```

```python
#Splitting the Training and Target Features
x=Encoded_Data.drop(['Name','Date of Admission', 'Doctor','Billing Amount','Room Number','Admission Type','Discharge Date', 'Medication','Test Results', 'Length of Stay'], axis=1)
y=Encoded_Data['Admission Type']
```

```python
print(x.shape,y.shape)
```

```python
#Training and Testing the Data
X_train, X_test, Y_train, Y_test = train_test_split(x,y, test_size=0.2, random_state=2)
```

```python
print(X_train.shape, X_test.shape, Y_train.shape, Y_test.shape)
```

```python
Models={
    'Logistic Regression':LogisticRegression(),
    'Decision Tree':DecisionTreeClassifier(),
    'Random Forest':RandomForestClassifier()}
```

# **Model Evaluation**

```python
#Creating an Empty Result Set
Results = []
```

```python
#Training and Evaluating the Each Model
for model_name, model in Models.items():
  model.fit(X_train, Y_train)
  Y_pred = model.predict(X_test)

  accuracy = accuracy_score(Y_test, Y_pred)
  precision = precision_score(Y_test, Y_pred, average='weighted',zero_division=0)
  recall = recall_score(Y_test, Y_pred, average='weighted',zero_division=0)
  f1 = f1_score(Y_test, Y_pred, average='weighted',zero_division=0)

  Results.append([model_name, accuracy,precision,recall,f1])
```

```python
# Create a DataFrame to store results
results_df = pd.DataFrame(Results, columns=['Model', 'Accuracy', 'Precision', 'Recall', 'F1-Score'])
results_df.head(3)
```
